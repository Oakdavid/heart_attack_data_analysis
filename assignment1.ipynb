# Advanced Cholesterol Distribution Plot:

```python
                                # 1

# Create a Cholesterol Distribution Plot (histogram + KDE curve)
# to analyze the distribution of cholesterol levels in the dataset.

sns.histplot(df[df['Heart_Attack_Outcome'] == 0]['Cholesterol_Level'], kde=True, color='green', bins=30, alpha=0.6, label='No Heart Attack')
sns.histplot(df[df['Heart_Attack_Outcome'] == 1]['Cholesterol_Level'], kde=True, color='red', bins=30, alpha=0.6, label='Heart Attack')
```

```python
#                               2.  

# Customize the plot with gridlines, labels, and a title.

plt.title('Cholesterol Level Distribution by Heart Attack Outcome', fontsize=14)
plt.xlabel('Cholesterol Level', fontsize=13)
plt.ylabel('Frequency', fontsize=13)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

Examine the plot and discuss insights related to cholesterol levels.

From the chart, we can see that people who had a heart attack usually have higher cholesterol levels than those who didn’t.
The red part of the chart (heart attack group) is mostly on the right side, meaning their cholesterol is higher.
The green part (no heart attack) is more on the left side, meaning their cholesterol is lower.
This shows that when a person’s cholesterol level is high, their chance of having a heart attack also increases.
But it’s not the only cause — some people with normal cholesterol still had heart attacks.
