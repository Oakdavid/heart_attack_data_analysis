# Advanced Cholesterol Distribution Plot:

```python
                                # 1

# Create a Cholesterol Distribution Plot (histogram + KDE curve)
# to analyze the distribution of cholesterol levels in the dataset.

sns.histplot(df[df['Heart_Attack_Outcome'] == 0]['Cholesterol_Level'], kde=True, color='green', bins=30, alpha=0.6, label='No Heart Attack')
sns.histplot(df[df['Heart_Attack_Outcome'] == 1]['Cholesterol_Level'], kde=True, color='red', bins=30, alpha=0.6, label='Heart Attack')
```

```python
#                               2.  

# Customize the plot with gridlines, labels, and a title.

plt.title('Cholesterol Level Distribution by Heart Attack Outcome', fontsize=14)
plt.xlabel('Cholesterol Level', fontsize=13)
plt.ylabel('Frequency', fontsize=13)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

                         3
Examine the plot and discuss insights related to cholesterol levels.

From the chart, we can see that people who had a heart attack usually have higher cholesterol levels than those who didn’t.
The red part of the chart (heart attack group) is mostly on the right side, meaning their cholesterol is higher.
The green part (no heart attack) is more on the left side, meaning their cholesterol is lower.
This shows that when a person’s cholesterol level is high, their chance of having a heart attack also increases.
But it’s not the only cause — some people with normal cholesterol still had heart attacks.


# Overlaying Line and Scatter Plots:


```python

#                                  1
#  Overlay a scatter plot and a line plot to show the 
# relationship between two variables (e.g., age vs. cholesterol levels).

# fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# Scatter and Line Plot Example
plt.figure(figsize=(8, 6))

# Scatter plot for Age vs Cholesterol_Level
sns.scatterplot(data=df, x="Age", y="Cholesterol_Level", color="blue", s=60, alpha=0.6, label="Data Points")

# Line plot (showing trend)
sns.lineplot(data=df, x="Age", y="Cholesterol_Level", color="red", linewidth=2, label="Trend Line")

# Add labels and title
plt.title("Relationship between Age and Cholesterol Levels", fontsize=14)
plt.xlabel("Age", fontsize=12)
plt.ylabel("Cholesterol Level", fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)

# Show plot
plt.show()
```

```python
#                                 2
# Customize the appearance of both plots 
#(e.g., color, size, transparency, etc.).

plt.figure(figsize=(8, 6))
sns.set_style("whitegrid")
sns.scatterplot(
    data=df, 
    x="Age", 
    y="Cholesterol_Level", 
    color="purple", 
    s=80, 
    alpha=0.7, 
    label="Data Points"
)
# line color with cutomization
sns.lineplot(
    data=df, 
    x="Age", 
    y="Cholesterol_Level", 
    color="orange", 
    linewidth=3, 
    label="Trend Line"
)

plt.title("Customized scatter and line plot: Age vs Cholesterol Level", fontsize=16)
plt.xlabel("Age", fontsize=14)
plt.ylabel("Cholesterol Level", fontsize=14)
plt.legend()
plt.show()
```

State your observations for the relationship between the variables based on the plots.

Observation
The scatter plot shows that as age increases, cholesterol levels tend to rise.
The line plot confirms a positive trend — older individuals generally have higher
cholesterol levels. Some outliers are visible, indicating variation among individuals


# Creating Multi-Axes Plots:


```python
# question 1
fig, ax = plt.subplots(1, 2, figsize=(14, 6))
plt.tight_layout(pad=5.0)
```

```python
# Question 2

# Create subplot (1 row, 2 columns)
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# ====== Bar Chart for Cholesterol Levels by Gender (First Axis) ======
sns.barplot(data=df, x="Gender", y="Cholesterol_Level", palette="Set2", ax=ax[0])

# Add title and labels for first plot
ax[0].set_title("Average Cholesterol Level by Gender", fontsize=14)
ax[0].set_xlabel("Gender", fontsize=12)
ax[0].set_ylabel("Cholesterol Level (mg/dL)", fontsize=12)


ax[1].set_visible(False)

# Show the figure
plt.tight_layout()
plt.show()
```

```python
# Question 3

# Create subplot (1 row, 2 columns)
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# ====== Bar Chart for Cholesterol Levels by Gender (First Axis) ======
sns.barplot(data=df, x="Gender", y="Cholesterol_Level", palette="Set2", ax=ax[0])
ax[0].set_title("Average Cholesterol Level by Gender", fontsize=14)
ax[0].set_xlabel("Gender", fontsize=12)
ax[0].set_ylabel("Cholesterol Level (mg/dL)", fontsize=12)

# ====== Scatter Plot for Blood Pressure vs Age (Second Axis) ======
sns.scatterplot(
    data=df,
    x="Age",
    y="Blood_Pressure_Diastolic",
    hue="Gender",        # colors by gender
    palette="coolwarm",
    s=70,
    alpha=0.7,
    ax=ax[1]
)

# Add title and labels for second plot
ax[1].set_title("Relationship between Age and Blood Pressure", fontsize=14)
ax[1].set_xlabel("Age (Years)", fontsize=12)
ax[1].set_ylabel("Blood Pressure (mmHg)", fontsize=12)
ax[1].legend(title="Gender")

# Adjust layout to prevent overlap
plt.tight_layout()

# Show both plots together
plt.show()
```
